Все выполненные доработки должны быть протестированы.

````python
# UnorderedList с лекции (файл взят у МОА потому что нам не скинули архив)

# 1
class Node:
    def __init__(self, initdata):
        self.data = initdata
        self.next = None

    def getData(self):
        return self.data

    def getNext(self):
        return self.next

    def setData(self, newdata):
        self.data = newdata

    def setNext(self, newnext):
        self.next = newnext


class UnorderedList:

    def __init__(self):
        self.head = None

    def isEmpty(self):
        return self.head == None

    def add(self, item):
        temp = Node(item)
        temp.setNext(self.head)
        self.head = temp

    def size(self):
        current = self.head
        count = 0
        while current != None:
            count = count + 1
            current = current.getNext()

        return count

    def search(self, item):
        current = self.head
        found = False
        while current != None and not found:
            if current.getData() == item:
                found = True
            else:
                current = current.getNext()

        return found

    def remove(self, item):
        current = self.head
        previous = None
        found = False
        while not found:
            if current.getData() == item:
                found = True
            else:
                previous = current
                current = current.getNext()

        if previous == None:
            self.head = current.getNext()
        else:
            previous.setNext(current.getNext())
            
    # добавляем методы по 1му пункту (append, index, inset, pop, pop(pos))
    def append(self, item):
        new_node = Node(item)
        if self.head is None:
            self.head = new_node
            return
        current = self.head
        while current.getNext() is not None:
            current = current.getNext()
        current.setNext(new_node)
    
    def index(self, item):
        current = self.head
        pos = 0
        while current is not None:
            if current.getData() == item:
                return pos
            current = current.getNext()
            pos += 1
        raise ValueError("Элемент не найден")
    
    def insert(self, pos, item):
        if pos == 0:
            self.add(item)
            return
        new_node = Node(item)
        current = self.head
        previous = None
        current_pos = 0
        while current is not None and current_pos < pos:
            previous = current
            current = current.getNext()
            current_pos += 1
        if previous is None:
            raise IndexError("Недопустимая позиция вставки")
        previous.setNext(new_node)
        new_node.setNext(current)
    
    def pop(self, pos=None):
        if self.head is None:
            raise IndexError("Попытка удалить из пустого списка")
        
        # для pop()
        if pos is None:
            current = self.head
            previous = None
            while current.getNext() is not None:
                previous = current
                current = current.getNext()
            if previous is None:
                self.head = None
            else:
                previous.setNext(None)
            return current.getData()
        
        # для pop(pos)
        current = self.head
        previous = None
        current_pos = 0
        while current is not None and current_pos < pos:
            previous = current
            current = current.getNext()
            current_pos += 1
        if current is None:
            raise IndexError("Индекс вне диапазона")
        if previous is None:
            self.head = current.getNext()
        else:
            previous.setNext(current.getNext())
        return current.getData()
    
    # 2
    def __str__(self):
        elements = []
        current = self.head
        while current is not None:
            elements.append(repr(current.getData()))
            current = current.getNext()
        return "[" + ", ".join(elements) + "]"
    
    # 3
    def slice(self, start, stop):
        if start < 0 or stop < 0 or start > stop:
            raise IndexError("Недопустимые границы среза")
        new_list = UnorderedList()
        current = self.head
        index = 0
        while current is not None and index < stop:
            if index >= start:
                new_list.append(current.getData())
            current = current.getNext()
            index += 1
        return new_list
    
lst = UnorderedList()
lst.add(32)
lst.add(22)
lst.add(15)
lst.append(41)
print(lst)
print(lst.index(3))
lst.insert(2, 5)
print(lst)
print(lst.pop())
print(lst.pop(1))
print(lst)
print(lst.slice(1, 3))
````

````python
# 1) итеративный
def reverse_iterative(head):
    previous = None
    current = head
    while current is not None:
        next_node = current.getNext()
        current.setNext(previous)
        previous = current
        current = next_node
    return previous

# 2) рекурсивный
def reverse_recursive(head):
    if head is None or head.getNext() is None:
        return head
    new_head = reverse_recursive(head.getNext())
    head.getNext().setNext(head)
    head.setNext(None)
    return new_head

# -------------------------------------------

# 1) итеративный

lst = UnorderedList()
lst.add(3)
lst.add(2)
lst.add(1)
print("Исходный список:", lst)

reversed_head_iter = reverse_iterative(lst.head)
print("Итеративно реверсирован:", end=" ")
temp_list = UnorderedList()
temp_list.head = reversed_head_iter
print(temp_list)

# 2) рекурсивный

lst = UnorderedList()
lst.add(3)
lst.add(2)
lst.add(1)  # [1, 2, 3]

reversed_head_rec = reverse_recursive(lst.head)
print("Рекурсивно реверсирован:", end=" ")
temp_list = UnorderedList()
temp_list.head = reversed_head_iter
print(temp_list)
````

````python
# стек
class LinkedListStack:
    def __init__(self):
        self.list = UnorderedList()

    def isEmpty(self):
        return self.list.isEmpty()

    def push(self, item):
        self.list.add(item)

    def pop(self):
        if self.list.isEmpty():
            raise IndexError("Удаляем из пустого стека")
        item = self.list.head.getData()
        self.list.remove(item)
        return item

    def peek(self):
        if self.list.isEmpty():
            raise IndexError("Смотрим из пустого стека")
        return self.list.head.getData()

    def size(self):
        return self.list.size()
    
    def __str__(self):
        current = self.list.head
        elements = []
        while current is not None:
            elements.append(repr(current.getData()))
            current = current.getNext()
        return "Stack(" + ", ".join(elements) + ")"

# очередь
class LinkedListQueue:
    def __init__(self):
        self.list = UnorderedList()

    def isEmpty(self):
        return self.list.isEmpty()

    def enqueue(self, item):
        self.list.append(item)

    def dequeue(self):
        if self.list.isEmpty():
            raise IndexError("Дек из пустой очереди")
        current = self.list.head
        data = current.getData()
        self.list.remove(data)
        return data

    def size(self):
        return self.list.size()
    
    def __str__(self):
        current = self.list.head
        elements = []
        while current is not None:
            elements.append(repr(current.getData()))
            current = current.getNext()
        return "Queue(" + ", ".join(elements) + ")"

# дек
class LinkedListDeque:
    def __init__(self):
        self.list = UnorderedList()

    def isEmpty(self):
        return self.list.isEmpty()

    def addFront(self, item):
        self.list.add(item)

    def addRear(self, item):
        self.list.append(item)

    def removeFront(self):
        if self.list.isEmpty():
            raise IndexError("Удаляем следующий из пустого дека")
        data = self.list.head.getData()
        self.list.remove(data)
        return data

    def removeRear(self):
        if self.list.isEmpty():
            raise IndexError("Удаляем предыдущий из пустого дека")
        current = self.list.head
        previous = None
        while current.getNext() is not None:
            previous = current
            current = current.getNext()
        if previous is None:
            self.list.head = None
        else:
            previous.setNext(None)
        return current.getData()

    def size(self):
        return self.list.size()
    
    def __str__(self):
        current = self.list.head
        elements = []
        while current is not None:
            elements.append(repr(current.getData()))
            current = current.getNext()
        return "Deque(" + ", ".join(elements) + ")"
    
# Тестирование с применением новых методов __str__

stack = LinkedListStack()
queue = LinkedListQueue()
deque = LinkedListDeque()

# Стек
stack.push(10)
stack.push(20)
stack.push(30)

print("Размер стека перед удалением элемента: ", stack.size())
print("Стек: ", stack)
stack.pop()
print("Размер стека после удаления элемента: ", stack.size())
print("Стек: ", stack)
print()
print()

# Очередь
queue.enqueue(1)
queue.enqueue(2)
queue.enqueue(3)

print("Размер очереди перед удалением элемента: ", queue.size())
print("Очередь: ", queue)
queue.dequeue()
print("Размер очереди после удаления элемента: ", queue.size())
print("Очередь: ", queue)
print()
print()

# Дек
deque.addFront('a')
deque.addRear('b')
deque.addFront('c')

print("Размер дека перед удалением переднего и заднего элементов: ", deque.size())
print("Дек: ", deque)
deque.removeFront()
deque.removeRear()
print("Размер дека после удалений: ", deque.size())
print("Дек: ", deque)

print()
print()
````

Реализуйте класс DoubleList для построения двухсвязных списков, в которых каждый узел содержит ссылку на узел, предшествующий ему в списке, и ссылку на следующий узел.

Класс должен иметь следующие методы: проверка списка на пустоту, поиск элемента, размер списка, вставка в начало, вставка в конец, удаление из начала, удаление из конца, вставка перед указанным узлом, вставка после указанного узла, удаление указанного узла.

Проведите тестирование методов реализованного класса.

````python
class DoubleNode:
    def __init__(self, data):
        self.data = data
        self.prev = None
        self.next = None

    def getData(self):
        return self.data

    def getPrev(self):
        return self.prev

    def getNext(self):
        return self.next

    def setData(self, data):
        self.data = data

    def setPrev(self, prev):
        self.prev = prev

    def setNext(self, next_node):
        self.next = next_node

class DoubleList:
    def __init__(self):
        self.head = None
        self.tail = None

    def isEmpty(self):
        return self.head is None

    def size(self):
        count = 0
        current = self.head
        while current is not None:
            count += 1
            current = current.getNext()
        return count

    def search(self, item):
        current = self.head
        while current is not None:
            if current.getData() == item:
                return True
            current = current.getNext()
        return False

    def add_front(self, item):
        new_node = DoubleNode(item)
        if self.isEmpty():
            self.head = new_node
            self.tail = new_node
        else:
            new_node.setNext(self.head)
            self.head.setPrev(new_node)
            self.head = new_node

    def add_rear(self, item):
        new_node = DoubleNode(item)
        if self.isEmpty():
            self.head = new_node
            self.tail = new_node
        else:
            new_node.setPrev(self.tail)
            self.tail.setNext(new_node)
            self.tail = new_node

    def remove_front(self):
        if self.isEmpty():
            raise IndexError("Удаление из пустого списка")
        removed_data = self.head.getData()
        if self.head == self.tail:
            self.head = None
            self.tail = None
        else:
            self.head = self.head.getNext()
            self.head.setPrev(None)
        return removed_data

    def remove_rear(self):
        if self.isEmpty():
            raise IndexError("Удаление из пустого списка")
        removed_data = self.tail.getData()
        if self.head == self.tail:
            self.head = None
            self.tail = None
        else:
            self.tail = self.tail.getPrev()
            self.tail.setNext(None)
        return removed_data

    def insert_before(self, ref_node_data, item):
        if self.isEmpty():
            raise ValueError("Список пуст")
        current = self.head
        while current is not None and current.getData() != ref_node_data:
            current = current.getNext()
        if current is None:
            raise ValueError("Указанный узел не найден")

        new_node = DoubleNode(item)
        prev_node = current.getPrev()

        new_node.setNext(current)
        new_node.setPrev(prev_node)
        current.setPrev(new_node)

        if prev_node is None:
            self.head = new_node
        else:
            prev_node.setNext(new_node)

    def insert_after(self, ref_node_data, item):
        if self.isEmpty():
            raise ValueError("Список пуст")
        current = self.head
        while current is not None and current.getData() != ref_node_data:
            current = current.getNext()
        if current is None:
            raise ValueError("Указанный узел не найден")

        new_node = DoubleNode(item)
        next_node = current.getNext()

        new_node.setPrev(current)
        new_node.setNext(next_node)
        current.setNext(new_node)

        if next_node is None:
            self.tail = new_node
        else:
            next_node.setPrev(new_node)

    def remove(self, item):
        if self.isEmpty():
            raise ValueError("Список пуст")
        current = self.head
        while current is not None and current.getData() != item:
            current = current.getNext()
        if current is None:
            raise ValueError("Указанный элемент не найден")

        prev_node = current.getPrev()
        next_node = current.getNext()

        if prev_node is None:
            self.head = next_node
        else:
            prev_node.setNext(next_node)

        if next_node is None:
            self.tail = prev_node
        else:
            next_node.setPrev(prev_node)

    def __str__(self):
        elems = []
        current = self.head
        while current is not None:
            elems.append(repr(current.getData()))
            current = current.getNext()
        return "[" + ", ".join(elems) + "]"
    
# тестим
dl = DoubleList()
print("Пустой список:", dl)
print("Проверка на пустоту:", dl.isEmpty())

dl.add_front(10)
dl.add_rear(20)
dl.add_front(5)
dl.add_rear(30)
print("После добавления в начало и конец:", dl)
print("Размер списка:", dl.size())  # 4
print("Поиск элемента 20:", dl.search(20))
print("Поиск элемента 100:", dl.search(100))

dl.insert_before(20, 15)
print("После вставки 15 перед 20:", dl)

dl.insert_after(20, 25)
print("После вставки 25 после 20:", dl)

print("Удалён из начала:", dl.remove_front())
print("Удалён из конца:", dl.remove_rear())
print("Список после удалений:", dl)

dl.remove(15)
print("После удаления 15:", dl)

try:
    dl.remove(100)
except ValueError as e:
    print("Ошибка удаления:", e)
````

Напишите программу, которая, используя структуру данных стек, проверит HTML документ на соответствие открывающих и закрывающих тегов.

За основу решения можно взять алгоритм проверки скобочной последовательности из лекционных материалов.

````python
class Stack:
    def __init__(self):
        self.items = []

    def isEmpty(self):
        return self.items == []

    def push(self, item):
        self.items.append(item)

    def pop(self):
        return self.items.pop()

    def peek(self):
        return self.items[-1]

    def size(self):
        return len(self.items)

def html_tag_checker(html_string):
    s = Stack()
    i = 0
    
    while i < len(html_string):
        if html_string[i] == '<':
            j = html_string.find('>', i + 1)
            if j == -1:
                return False
            
            tag = html_string[i+1:j].strip()
            if not tag:
                return False
            
            if tag[0] != '/':
                tag_name = tag.split()[0]
                s.push(tag_name)
                
            else:
                tag_name = tag[1:].split()[0]
                if s.isEmpty():
                    return False
                
                top_tag = s.pop()
                if top_tag != tag_name:
                    return False
            i = j
        i += 1
    return s.isEmpty()


# тестим, вернёт истину
html_code = """
<html>
    <head>
        <title>Example</title>
    </head>
    <body>
        <h1>Hello, world</h1>
    </body>
</html>
"""

print(html_tag_checker(html_code))

# тестим, вернёт ложь
html_code_unbalanced = """
<html>
    <head><title>Example</title>
    </head>
    <body>
        <h1>Hello, world</h2>
    </body>
</html>
"""

print(html_tag_checker(html_code_unbalanced))
````

Стековым он называется так как действует по следующим правилам: если нам встречается число - мы должны поместить его в стек; если знак операции - мы должны достать из стека столько чисел, сколько необходимо для выполнения данной операции и выполнить ее.

Именно так и работает т.н. математический сопроцессор (например, в INtel8086+AMD64)- часть ЦП, отвечающая за операции над числами с плавающей запятой.

Реализуйте стековый калькулятор. Напишите программу, которая читает выражение в обратной польской нотации и выводит получившееся значение или пишет, что выражение составлено не корректно (если оно некорректно).

````python
def rpn_calculator(expression):
    stack = []
    operators = {'+', '-', '*', '/'}

    for token in expression.split():
        if token.isdigit() or (token.startswith('-') and token[1:].isdigit()):
            stack.append(float(token))
        elif token in operators:
            if len(stack) < 2:
                return "Некорректное выражение"
            b = stack.pop()
            a = stack.pop()
            if token == '+':
                stack.append(a + b)
            elif token == '-':
                stack.append(a - b)
            elif token == '*':
                stack.append(a * b)
            elif token == '/':
                if b == 0:
                    return "Деление на ноль"
                stack.append(a / b)
        else:
            return "Некорректное выражение"

    if len(stack) != 1:
        return "Некорректное выражение"

    return stack[0]

example = "3 4 + 2 * 7 /"
result = rpn_calculator(example)
print(result)
````
