В примере алгоритма поиска в ширину из лекции (https://cloud.mail.ru/public/9yCy/LitbGAfEg) структура данных граф представлена словарем. Переделайте этот пример, заменив словарь на реализацию класса Graph из
лекционных материалов.

```python
# код по ссылке:
from collections import deque

def person_is_seller(name):
      return name[0] == 'a'

graph_dict = {}
graph_dict["you"] = ["alice", "bob", "claire"]
graph_dict["bob"] = ["anuj", "peggy"]
graph_dict["alice"] = ["peggy"]
graph_dict["claire"] = ["thom", "jonny"]
graph_dict["anuj"] = []
graph_dict["peggy"] = []
graph_dict["thom"] = []
graph_dict["jonny"] = []

def search(name):
    search_queue = deque()
    search_queue += graph_dict[name]
    searched = set()
    while search_queue:
        person = search_queue.popleft()
        if person not in searched:
            if person_is_seller(person):
                print(person + " is a mango seller!")
                return True
            else:
                search_queue += graph_dict[person]
                searched.add(person)
    return False

search("you")
```

```python
# из лекционных материалов:
class Vertex:
    def __init__(self, key):
        self.id = key
        self.connectedTo = {}

    def addNeighbor(self, nbr, weight=0):
        self.connectedTo[nbr] = weight

    def __str__(self):
        return str(self.id) + ' connectedTo: ' + str([x.id for x in self.connectedTo])

    def getConnections(self):
        return self.connectedTo.keys()

    def getId(self):
        return self.id

    def getWeight(self, nbr):
        return self.connectedTo[nbr]


class Graph:
    def __init__(self):
        self.vertList = {}
        self.numVertices = 0

    def addVertex(self, key):
        self.numVertices = self.numVertices + 1
        newVertex = Vertex(key)
        self.vertList[key] = newVertex
        return newVertex

    def getVertex(self, n):
        if n in self.vertList:
            return self.vertList[n]
        else:
            return None

    def __contains__(self, n):
        return n in self.vertList

    def addEdge(self, f, t, cost=0):
        if f not in self.vertList:
            self.addVertex(f)
        if t not in self.vertList:
            self.addVertex(t)
        self.vertList[f].addNeighbor(self.vertList[t], cost)

    def getVertices(self):
        return self.vertList.keys()

    def __iter__(self):
        return iter(self.vertList.values())

graph = Graph()
# graph.addVertex("you")
# graph.addVertex("bob")
# graph.addVertex("alice")
# graph.addVertex("claire")
# graph.addVertex("anuj")
# graph.addVertex("peggy")
# graph.addVertex("thom")
# graph.addVertex("jonny")
# Не добавляем узлы руками т.к. все они создаются при создании вершин

graph.addEdge("you", "alice")
graph.addEdge("you", "bob")
graph.addEdge("you", "claire")

graph.addEdge("bob", "anuj")
graph.addEdge("bob", "peggy")

graph.addEdge("alice", "peggy")

graph.addEdge("claire", "thom")
graph.addEdge("claire", "jonny")

vertices = graph.getVertices()
print(f"Вершины: {vertices}")
for vert in graph:
    for neighbor in vert.getConnections():
        print("(%s, %s)" % (vert.getId(), neighbor.getId()))
print("\n")
        
search("you")
```

![task2_example.png](task2_example.png)

```python
def reverseGraph(graph):
    reversed_graph = Graph()
    for vertex_key in graph.getVertices():
        reversed_graph.addVertex(vertex_key)
    for vertex in graph:
        from_vert = vertex.getId()
        for neighbor in vertex.getConnections():
            to_vert = neighbor.getId()
            weight = vertex.getWeight(neighbor)
            
            reversed_graph.addEdge(to_vert, from_vert, weight)
    return reversed_graph
        
graph_2 = Graph()

# Изначальный
original_graph = Graph()
original_graph.addEdge(0, 1)
original_graph.addEdge(0, 2)
original_graph.addEdge(2, 1)
original_graph.addEdge(1, 3)
print(f"Изначальный граф, его связи: ")
for vert in original_graph:
    for neighbor in vert.getConnections():
        print("(%s, %s)" % (vert.getId(), neighbor.getId()))
print("\n")

# Обращённый
reversed_graph = reverseGraph(original_graph)
for vert in reversed_graph:
    for neighbor in vert.getConnections():
        print("(%s, %s)" % (vert.getId(), neighbor.getId()))
```
