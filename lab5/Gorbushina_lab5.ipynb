```python
# код по ссылке:
from collections import deque

def person_is_seller(name):
      return name[0] == 'a'

graph_dict = {}
graph_dict["you"] = ["alice", "bob", "claire"]
graph_dict["bob"] = ["anuj", "peggy"]
graph_dict["alice"] = ["peggy"]
graph_dict["claire"] = ["thom", "jonny"]
graph_dict["anuj"] = []
graph_dict["peggy"] = []
graph_dict["thom"] = []
graph_dict["jonny"] = []

def search(name):
    search_queue = deque()
    search_queue += graph_dict[name]
    searched = set()
    while search_queue:
        person = search_queue.popleft()
        if person not in searched:
            if person_is_seller(person):
                print(person + " is a mango seller!")
                return True
            else:
                search_queue += graph_dict[person]
                searched.add(person)
    return False

search("you")
```

```python
# из лекционных материалов:
class Vertex:
    def __init__(self, key):
        self.id = key
        self.connectedTo = {}

    def addNeighbor(self, nbr, weight=0):
        self.connectedTo[nbr] = weight

    def __str__(self):
        return str(self.id) + ' connectedTo: ' + str([x.id for x in self.connectedTo])

    def getConnections(self):
        return self.connectedTo.keys()

    def getId(self):
        return self.id

    def getWeight(self, nbr):
        return self.connectedTo[nbr]


class Graph:
    def __init__(self):
        self.vertList = {}
        self.numVertices = 0

    def addVertex(self, key):
        self.numVertices = self.numVertices + 1
        newVertex = Vertex(key)
        self.vertList[key] = newVertex
        return newVertex

    def getVertex(self, n):
        if n in self.vertList:
            return self.vertList[n]
        else:
            return None

    def __contains__(self, n):
        return n in self.vertList

    def addEdge(self, f, t, cost=0):
        if f not in self.vertList:
            self.addVertex(f)
        if t not in self.vertList:
            self.addVertex(t)
        self.vertList[f].addNeighbor(self.vertList[t], cost)

    def getVertices(self):
        return self.vertList.keys()

    def __iter__(self):
        return iter(self.vertList.values())

graph = Graph()
# graph.addVertex("you")
# graph.addVertex("bob")
# graph.addVertex("alice")
# graph.addVertex("claire")
# graph.addVertex("anuj")
# graph.addVertex("peggy")
# graph.addVertex("thom")
# graph.addVertex("jonny")
# Не добавляем узлы руками т.к. все они создаются при создании вершин

graph.addEdge("you", "alice")
graph.addEdge("you", "bob")
graph.addEdge("you", "claire")

graph.addEdge("bob", "anuj")
graph.addEdge("bob", "peggy")

graph.addEdge("alice", "peggy")

graph.addEdge("claire", "thom")
graph.addEdge("claire", "jonny")

vertices = graph.getVertices()
print(f"Вершины: {vertices}")
for vert in graph:
    for neighbor in vert.getConnections():
        print("(%s, %s)" % (vert.getId(), neighbor.getId()))
print("\n")
        
search("you")
```

![task2_example.png](task2_example.png)

```python
def reverseGraph(graph):
    reversed_graph = Graph()
    for vertex_key in graph.getVertices():
        reversed_graph.addVertex(vertex_key)
    for vertex in graph:
        from_vert = vertex.getId()
        for neighbor in vertex.getConnections():
            to_vert = neighbor.getId()
            weight = vertex.getWeight(neighbor)
            
            reversed_graph.addEdge(to_vert, from_vert, weight)
    return reversed_graph
        
graph_2 = Graph()

# Изначальный
original_graph = Graph()
original_graph.addEdge(0, 1)
original_graph.addEdge(0, 2)
original_graph.addEdge(2, 1)
original_graph.addEdge(1, 3)
print(f"Изначальный граф, его связи: ")
for vert in original_graph:
    for neighbor in vert.getConnections():
        print("(%s, %s)" % (vert.getId(), neighbor.getId()))
print("\n")

# Обращённый
reversed_graph = reverseGraph(original_graph)
for vert in reversed_graph:
    for neighbor in vert.getConnections():
        print("(%s, %s)" % (vert.getId(), neighbor.getId()))
```

Напишите программу, которая при помощи поиска в глубину (Depth-first search, DFS) проведет топологическую сортировку представленного графа и выдаст вариант корректной последовательности шагов. Например:

Используйте класс Graph из лекционных материалов.

```python
def pancakes_graph():
    graph = Graph()
    # по первой картинке, абзацы по вершинам
    graph.addEdge("3/4 cup milk", "1 cup mix")
    graph.addEdge("1 egg", "1 cup mix")
    graph.addEdge("1 Tbl Oil", "1 cup mix")
    
    graph.addEdge("1 cup mix", "heat syrup")
    graph.addEdge("1 cup mix", "pour 1/4 cup")
    
    graph.addEdge("heat griddle", "pour 1/4 cup")
    
    graph.addEdge("pour 1/4 cup", "turn when bubbly")
    
    graph.addEdge("turn when bubbly", "eat")
    graph.addEdge("heat syrup", "eat")
    
    return graph

def depth_first_search(graph: Graph):
    visited = set()
    ordered_list = []
    
    def dfs_for_vertex(vertex: Vertex):
        visited.add(vertex)
        for next_vertex in vertex.getConnections():
            if next_vertex not in visited:
                dfs_for_vertex(next_vertex)
        ordered_list.insert(0, vertex.getId())
        
    for vertex in graph:
        if vertex not in visited:
            dfs_for_vertex(vertex)
            
    return ordered_list
        
def task_3():
    graph = pancakes_graph()
    result = depth_first_search(graph)
    for step in result:
        print(step)
        
task_3()
```

```python
def cities_graph():
    routes = Graph()
    
    routes.addEdge("Киселевск", "Таштагол", 214)
    routes.addEdge("Киселевск", "Прокопьевск", 19)
    routes.addEdge("Прокопьевск", "Междуреченск", 117)
    routes.addEdge("Прокопьевск", "Белово", 87)
    routes.addEdge("Осинники", "Красноярск", 767)
    routes.addEdge("Таштагол", "Осинники", 148)
    routes.addEdge("Таштагол", "Красноярск", 896)
    routes.addEdge("Таштагол", "Междуреченск", 234)
    routes.addEdge("Междуреченск", "Красноярск", 830)
    routes.addEdge("Белово", "Междуреченск", 188)
    routes.addEdge("Белово", "Красноярск", 643)
    
    return routes

def find_shortest_path(graph: Graph, from_city: str, to_city: str):
    start_vertex = graph.getVertex(from_city)
    end_vertex = graph.getVertex(to_city)
    
    if not start_vertex or not end_vertex:
        print("В какой-то из этих городов не ездим :(")
        return
    
    # создаём словарь вершина: расстояние,
    # значение по умолчанию = беск (типа неопределено)
    distances = {vert: float('inf') for vert in graph}
    distances[start_vertex] = 0
    
    # словарь чтобы отмотать путь назад
    before = {vert: None for vert in graph}
    
    # список непосещённых вершин (сначала все)
    unvisited_vertices = {vert for vert in graph}
    
    while unvisited_vertices:
        current_vert: Vertex = None
        min_dist = float('inf')
        for vert in unvisited_vertices:
            if distances[vert] < min_dist:
                min_dist = distances[vert]
                current_vert = vert
                
        # больше нет вершин - бай-бай
        if current_vert is None:
            break
        
        if current_vert == end_vertex:
            break                        # ура мы доехали
        
        unvisited_vertices.remove(current_vert) # посмотрели все пути от вершины
        
        for neighbor in current_vert.getConnections():
            weight = current_vert.getWeight(neighbor)
            new_distance = distances[current_vert] + weight
            
            if new_distance < distances[neighbor]:
                distances[neighbor] = new_distance
                before[neighbor] = current_vert
                
    final_distance = distances[end_vertex]

    if final_distance == float('inf'):
        print("Пути между городами нет.")
    else:
        path = []
        current = end_vertex
        while current is not None:
            path.append(current.getId())
            current = before[current]
        path.reverse()
        
        path_string = " -> ".join(path)
        print(f"Путь: {path_string} -> Приехали!")
        print(f"Километров: {final_distance}")

def task5():
    print("Список доступных городов:\n - Киселевск\n - Таштагол\n - Прокопьевск \n - Белово\n - Осинники\n - Междуреченск \n - Красноярск")
    print("Введите город отправления:")
    from_city = input()
    print("Введите город назначения: ")
    to_city = input()
    graph = cities_graph()
    find_shortest_path(graph, from_city, to_city)
    
task5()
```
